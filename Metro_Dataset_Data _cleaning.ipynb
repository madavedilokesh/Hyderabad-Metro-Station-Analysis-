#Fill the Null Values With Median/Mean:-

```python
df['ticket_revenue_(inr)'] = df['ticket_revenue_(inr)'].astype(object)
```

```python
outliers_col = ['passenger_entries', 'passenger_exits','total_passenger_movement',
                'train_delays','complaints','lost_&_found_cases','station_rating']
outliers_col
```

```python
for oc in outliers_col:
    df[oc] = df[oc].fillna(df.groupby('station_name')[oc].transform('median'))
    df[oc] = df[oc].fillna(df[oc].median())
```

```python
no_outliers_col = ['sequence', 'gates_/_exits', 'lifts', 'escalators', 'stairs', 'hour',
                   'average_waiting_time_(min)', 'average_trip_time_(min)','train_frequency_(min)',
                   'temperature_(°c)']
no_outliers_col
```

```python
for noc in no_outliers_col:
  df[noc] = df[noc].fillna(df.groupby('station_name')[noc].transform('mean'))
  df[noc] = df[noc].fillna(df[noc].mean())
```

```python
df.isna().sum()
```

##Fill the Null Values With Mode:-

```python
df['ticket_revenue_(inr)'] = df['ticket_revenue_(inr)'].astype(float)
```

```python
cat_cols = df.select_dtypes(object).columns
cat_cols
```

```python
cat_col = ['drinking_water', 'parking', 'bus_connections', 'maintenance_issues']
cat_col
```

```python
for cc in cat_col:
    df[cc] = df[cc].fillna(df.groupby('station_name')[cc].transform(lambda x: x.mode()[0] if not x.mode().empty else None))
    df[cc] = df[cc].fillna(df[cc].mode()[0])
```

```python
df.isna().sum()
```

```python
df.shape
```

#Outliers:-

```python
Q3=df['ticket_revenue_(inr)'].quantile(0.75)
Q1=df['ticket_revenue_(inr)'].quantile(0.25)
Q3,Q1
```

```python
IQR=Q3-Q1
IQR
```

```python
upper_limit=Q3+1.5*IQR
lower_limit=Q1-1.5*IQR
upper_limit,lower_limit
```

```python
df = df[(df['ticket_revenue_(inr)']>=lower_limit) & (df['ticket_revenue_(inr)']<=upper_limit)]
```

#EDA

##1. Which metro stations have the highest passenger movement?

```python
df.columns
```

```python
top_stations = (df.groupby("station_name")["total_passenger_movement"].sum().nlargest(10).sort_values())
```

```python
plt.figure(figsize=(14,6))
sns.barplot(y=top_stations.values,x=top_stations.index,palette='Set2');
plt.title("Top 10 Metro Stations by Passenger Movement")
plt.xlabel("Total Passenger Movement")
plt.ylabel("Station Name")
plt.show()
```

##2. What is the proportion of weekday vs weekend passenger movement?

```python
day_movement = df.groupby("day_type")["total_passenger_movement"].sum()
```

```python
plt.figure(figsize=(6,6))
plt.pie(day_movement,labels=day_movement.index,autopct="%1.1f%%")
plt.title("Passenger Movement: Weekday vs Weekend")
plt.show()
```

##3. Is higher passenger movement associated with higher total revenue?

```python
passenger_revenue = df.groupby("line")[["total_passenger_movement","ticket_revenue_(inr)"]].sum()
```

```python
plt.figure(figsize=(6,6))
sns.scatterplot(data=passenger_revenue,x="total_passenger_movement",y="ticket_revenue_(inr)",s=150)
plt.xlabel("Total Passenger Movement")
plt.ylabel("Ticket Revenue (INR)")
plt.title("Passenger Movement vs Ticket Revenue")
plt.show()
```

##4. How frequently do different train-delay levels occur?


```python
plt.figure(figsize=(9,6))
sns.countplot(data=df,x="train_delays")
plt.title("Frequency of Train Delays")
plt.xlabel("Number of Train Delays")
plt.ylabel("Number of Records")
plt.show()
```

##5.  How does passenger movement vary under different weather conditions?

```python
weather_movement = (df.groupby("weather")["total_passenger_movement"].mean())
plt.figure(figsize=(8,5))
sns.barplot(y=weather_movement.values,x=weather_movement.index)
plt.title("Average Passenger Movement by Weather")
plt.xlabel("Weather")
plt.ylabel("Average Passenger Movement")
plt.xticks(rotation=0)
plt.show()
```

##8. How does ticket revenue change throughout the day?

```python
hourly_revenue = (df.groupby("hour")["ticket_revenue_(inr)"].mean())
plt.figure(figsize=(10,6))
plt.plot(hourly_revenue.index,hourly_revenue.values,marker="o")
plt.title("Average Ticket Revenue by Hour")
plt.xlabel("Hour")
plt.ylabel("Average Ticket Revenue (INR)")
plt.show()
```
